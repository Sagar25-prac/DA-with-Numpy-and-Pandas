**Multi Array**

```python
t = np.linspace(0,10,100)
x = np.cos(t)
y = np.sin(t)

z = t/2
```

```python
traj = np.column_stack((x,y,z))
traj
```

```python
print('Trajectory Shape: ', traj.shape)
print('Time Space: ', t.shape)
print('Trajectory Size: ' ,traj.size)
print('Time Size: ',t.size)
print('Trajectory Data Type: ',traj.dtype)
print('Time Data Type: ',t.dtype)
print('Trajectory Dimension: ',traj.ndim)
print('Time Dimension: ',t.ndim)
```

**Masking**

```python
x = np.array([-2,-1,1,2,2.5])

mask = x>0
print(x[mask])
print(x[x>0])

mask = x%2 ==0
print(x[mask])
```

```python
t = np.linspace(0,10,100)
x = np.cos(t)
y = np.sin(t)

z = t/2

traj = np.zeros((100,3),dtype=np.float64)
traj[:,0] = x
traj[:,1] = y
traj[:,2] = z

z_threashold = 3.0
mask = traj[:,2] < z_threashold
low_altitude_points = traj[mask,:]


fig = plt.figure(figsize =(8,6))
ax = fig.add_subplot(111,projection ='3d')
ax.plot(x,y,z,'b-',linewidth = 2)
ax.scatter(low_altitude_points[:,0],low_altitude_points[:,1],low_altitude_points[:,2],c='red',s=50,label = 'low altitude')
ax.set_xlabel('km')
ax.set_ylabel('km')
ax.set_zlabel('km')

plt.show()
```
