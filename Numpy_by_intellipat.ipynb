**Array Reshaping -> Reshape,Ravel and flatten**

```python
# Reshape

arr = np.array([1,2,3,4,5,6])
print(arr)

reshape = arr.reshape(2,3)
print(reshape)

reshaped = reshape.reshape(3,2)
print(reshaped)
```

```python
# Ravel -> used to covert into 1D array create a view
ravel = reshaped.ravel()
print(ravel)
print(reshaped)
```

```python
# flatten -> used to convert into 1D array but create a copy

flat = reshaped.flatten()
print(flat)
print(reshaped)
```

**Vectorization**

np.vector() -> convert a regular function to be applied on an array
